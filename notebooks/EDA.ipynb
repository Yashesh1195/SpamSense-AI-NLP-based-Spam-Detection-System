# Exploratory Data Analysis

Inspect the SMS spam dataset before modeling.

```python
from pathlib import Path
import sys

import pandas as pd
import plotly.express as px

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

DATA_PATH = PROJECT_ROOT / 'data' / 'processed' / 'SMSSpamCollection.txt'
messages = pd.read_csv(DATA_PATH, sep='\t', names=['label', 'message'])
messages['message_length'] = messages['message'].astype(str).str.len()
messages['word_count'] = messages['message'].astype(str).str.split().str.len()
messages.head()

class_counts = messages['label'].value_counts().reset_index()
class_counts.columns = ['label', 'count']
print(messages.info())
print(class_counts)
px.pie(class_counts, names='label', values='count', hole=0.45, title='Spam vs Ham Distribution')
px.histogram(messages, x='message_length', color='label', nbins=30, title='Message Length by Label')
```
